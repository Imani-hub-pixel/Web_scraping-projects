```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
Images_id=[]
Title=[]
Price=[]
Rating=[]
Status=[]

for page in range(1, 51):
    url='https://books.toscrape.com/catalogue/page-{}.html'.format(page)
    header={
        'User-Agent':'Mozilla/5.0'
    }
    page=requests.get(url,headers=header)
    page.encoding='utf-8'
    soup=BeautifulSoup(page.text,'html')
   
    books=soup.find_all('article',class_='product_pod')
    for book in books:
        Image_id=book.find('div',class_='image_container').find('img')['src']
        title=book.find('h3').find('a')['title']
        price=book.find('p',class_='price_color').text
        rating=book.find('p',class_='star-rating')['class'][1]
        status=book.find('p',class_='instock availability').text.strip()
        
        Images_id.append(Image_id)
        Title.append(title)
        Price.append(price)
        Rating.append(rating)
        Status.append(status)

        data={
        'Image_id':Images_id,
        'Title':Title,
        'Price':Price,
        'Rating':Rating,
        'Status':Status
        }
            
```

```python
df=pd.DataFrame(data)
df.to_csv('Books_to_scrape.csv',index=False)
```
